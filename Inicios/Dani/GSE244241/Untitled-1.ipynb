```python
import pandas as pd

# 1. Definir el nombre del archivo de entrada y salida
archivo_entrada = 'GSE244241_HBC_gv.DE (1).csv.gz'
archivo_salida = 'GSE244241_HBC_gv_procesado.csv'

try:
    # 2. Leer el archivo
    # Si el archivo está separado por comas:
    df = pd.read_csv(archivo_entrada)
    
    # Nota: Si el archivo estuviera separado por punto y coma (;) o tabulaciones, 
    # deberías usar sep=';' o sep='\t' respectivamente:
    # df = pd.read_csv(archivo_entrada, sep='\t')

    # 3. Mostrar la información básica para verificar que se leyó bien
    print("Primeras 5 filas del archivo:")
    print(df.head())
    print("\nInformación de las columnas:")
    print(df.info())

    # 4. Procesamiento básico (Opcional)
    # Por ejemplo, eliminar filas que estén completamente vacías
    df = df.dropna(how='all')
    
    # 5. Guardar como un CSV limpio y estandarizado
    # index=False evita que se guarde la columna de numeración de pandas
    df.to_csv(archivo_salida, index=False, encoding='utf-8')
    
    print(f"\n¡Archivo procesado y guardado con éxito como '{archivo_salida}'!")

except FileNotFoundError:
    print(f"Error: No se encontró el archivo '{archivo_entrada}'. Asegúrate de que esté en la misma carpeta que este script.")
except Exception as e:
    print(f"Ocurrió un error al procesar el archivo: {e}")
```

```python
# 1. Cargar el archivo original
archivo = 'GSE244241_HBC_gv_procesado.csv'
df = pd.read_csv(archivo)

# 2. Borrar la columna "Unnamed: 0"
if 'Unnamed: 0' in df.columns:
    df = df.drop(columns=['Unnamed: 0'])

# 3. Pasar "SYMBOLS" a la primera posición
if 'SYMBOLS' in df.columns:
    # Armamos una nueva lista de columnas con SYMBOLS primero
    nuevas_columnas = ['SYMBOLS'] + [col for col in df.columns if col != 'SYMBOLS']
    # Reordenamos el DataFrame
    df = df[nuevas_columnas]

# 4. Guardar el archivo limpio sin el índice de pandas
archivo_salida = 'GSE244241_HBC_gv_limpio.csv'
df.to_csv(archivo_salida, index=False)

print("¡Listo! Archivo procesado y guardado.")
```

```python
# 1. Cargar el archivo original
archivo = 'GSE244241_HBC.csv'
df = pd.read_csv(archivo)

# 2. Definir el umbral de significancia para el p-valor ajustado
umbral_adj_p = 0.05

# 3. Filtrar la tabla para quedarnos solo con los significativos
# Usamos el nombre exacto de la columna que me pasaste: 'adj.P.Val'
genes_significativos = df[df['adj.P.Val'] < umbral_adj_p]

# 4. Ver cuántos pasaron el filtro
cantidad_total = len(df)
cantidad_significativos = len(genes_significativos)

print(f"Total de genes analizados: {cantidad_total}")
print(f"Genes verdaderamente significativos (adj.P.Val < {umbral_adj_p}): {cantidad_significativos}")

# 5. Si hay genes significativos, los limpiamos y guardamos en un CSV nuevo
if cantidad_significativos > 0:
    # Aplicamos la limpieza que querías antes
    if 'Unnamed: 0' in genes_significativos.columns:
        genes_significativos = genes_significativos.drop(columns=['Unnamed: 0'])
    
    if 'SYMBOLS' in genes_significativos.columns:
        nuevas_cols = ['SYMBOLS'] + [c for c in genes_significativos.columns if c != 'SYMBOLS']
        genes_significativos = genes_significativos[nuevas_cols]
        
    archivo_salida = 'GSE244241_HBC_gv_Significativos.csv'
    genes_significativos.to_csv(archivo_salida, index=False)
    print(f"\n¡Éxito! La lista de los {cantidad_significativos} genes se guardó en '{archivo_salida}'.")
```
