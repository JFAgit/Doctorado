# Filtrar fail counts

```r
# Leer el archivo CSV
data <- read.csv("datacruda.csv")

# Filtrar las filas donde 'fail.count' es menor o igual a 0
data_filtrada <- subset(data, fail.count <= 0)

# Mostrar las primeras filas para verificar
head(data_filtrada)

# (Opcional) Guardar el nuevo dataset filtrado en un archivo
write.csv(data_filtrada, "datacruda_V0.1.csv", row.names = FALSE)
```

## Aplico base logaritmica

```r
# Leer correctamente los datos con coma decimal
data_vlog <- read.csv2("datacruda_V0.1.csv")
```

```r
names(data_vlog)
```

```r
# Leer el archivo como texto completo
raw_lines <- readLines("datacruda_V0.1.csv")

# Reemplazar las comas decimales por puntos (solo en los valores, no en los separadores de columnas)
# Esto asume que las columnas están separadas por coma, y que las comas decimales aparecen dentro de comillas
raw_lines <- gsub('([0-9]),([0-9])', '\\1.\\2', raw_lines)

# Escribir el archivo temporal corregido
writeLines(raw_lines, "datacruda_V0.1_tmp.csv")

# Ahora leer el archivo corregido normalmente
data_vlog <- read.csv("datacruda_V0.1_tmp.csv", stringsAsFactors = FALSE)

# Verificar columnas disponibles
names(data_vlog)[1:10]
```

```r
cols_log <- c(
  "WT_1.gProcessedSignal",
  "WT_2.gProcessedSignal",
  "Clon_4.gProcessedSignal",
  "Clon_10.gProcessedSignal",
  "RNA7_FN1_1.gProcessedSignal",
  "RNA7_FN1_2.gProcessedSignal",
  "RNA5_TLR4_1.gProcessedSignal",
  "RNA5_TLR4_2.gProcessedSignal"
)

data_vlog[cols_log] <- lapply(data_vlog[cols_log], function(x) log2(as.numeric(x) + 1))
```

```r
write.csv(data_vlog, "datacruda_V0.1_log2.csv", row.names = FALSE)
```

# Filtrado por cuantiles

```r
# Columnas a normalizar
cols_log <- c
  "WT_1.gProcessedSignal",
  "WT_2.gProcessedSignal",
  "Clon_4.gProcessedSignal",
  "Clon_10.gProcessedSignal",
  "RNA7_FN1_1.gProcessedSignal",
  "RNA7_FN1_2.gProcessedSignal",
  "RNA5_TLR4_1.gProcessedSignal",
  "RNA5_TLR4_2.gProcessedSignal"
)

# Extraer matriz numérica
matriz_signales <- as.matrix(data_vlog[cols_log])

# Normalización por cuantiles
matriz_normalizada <- normalize.quantiles(matriz_signales)

# Reasignar la matriz normalizada al dataframe
data_vlog[cols_log] <- matriz_normalizada

# Guardar archivo
write.csv(data_vlog, "datacruda_V0.1_log2_quantilenorm.csv", row.names = FALSE)
```
